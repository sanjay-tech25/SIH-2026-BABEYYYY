### Platform Curriculum Review & Theoretical Foundations
In Chapter 3, you discovered the single-qubit Pauli family: $X$ flips bits, $Z$ flips the relative phase of $|1\rangle$, $Y = iXZ$ does both, and phase gates $S = Z^{1/2}, T = Z^{1/4}$ rotate about the Z-axis by $\pi/2$ and $\pi/4$. Let's demonstrate how a phase flip $Z$ converts $|+\rangle$ into $|-\rangle$!

**Mathematical Formulation:**
$$S = \begin{bmatrix} 1 & 0 \\ 0 & i \end{bmatrix},   T = \begin{bmatrix} 1 & 0 \\ 0 & e^{i\pi/4} \end{bmatrix}$$
*Clifford Phase gate S and non-Clifford T gate unitary matrices.*

### Section 2: Reference Implementation & Theoretical Walkthrough
Analyze and execute the verified baseline implementation below. Observe how each operation transforms the state vector in Hilbert space.

```python
# Reference Implementation: Pauli Operators (X, Y, Z) & Phase Shifting (S, T)
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator

qc = QuantumCircuit(1, 1)
qc.h(0)
qc.s(0)
qc.s(0)
qc.h(0)
qc.measure(0, 0)

sim = AerSimulator()
counts = sim.run(transpile(qc, sim), shots=1024).result().get_counts()
print('H-S-S-H Interference counts:', counts)

# Display circuit structure:
try:
    print('\n--- Circuit Structure ---')
    print(qc.draw('text'))
except Exception as e:
    pass
```
