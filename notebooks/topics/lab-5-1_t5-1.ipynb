### 📖 What You Learned in the Platform Lesson
Quantum teleportation is the bedrock of distributed quantum computing and the Quantum Internet. Because of the No-Cloning Theorem, unknown quantum information cannot be copied. Teleportation transfers the complete quantum state $|\psi\rangle = \alpha|0\rangle + \beta|1\rangle$ from Qubit 0 to Qubit 2 using shared entanglement and 2 classical bits.

**Mathematical Formulation:**
$$|\psi\rangle_{\text{Bob}} = Z^{m_0} X^{m_1} |\psi\rangle_{\text{recovered}}$$
*Bob's conditional unitary correction recovering original quantum state.*

### 🧪 Step 2: Interactive Concept Walkthrough (Reference Demonstration)
Inspect and run the reference code below. Notice how each line executes a physical transformation in Hilbert space.

```python
# Reference Implementation for: Quantum State Teleportation Protocol
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator

qc = QuantumCircuit(3, 2)
qc.x(0)  # Teleport state |1>
qc.h(1); qc.cx(1, 2)  # Bell pair
qc.cx(0, 1); qc.h(0)  # Alice measurement
qc.measure([0, 1], [0, 1])

sim = AerSimulator()
print(qc.draw(output='text'))

# Display circuit structure:
try:
    print('\n--- Circuit Diagram ---')
    print(qc.draw('text'))
except Exception as e:
    pass
```

#### 💡 Supported Writing Styles & Innovation
In quantum programming, there is never only one way to write a circuit! Qiskit supports multiple styles:
- **Style A (Standard Shorthand)**: `qc = QuantumCircuit(3, 3); qc.h(0)`
- **Style B (Named Registers)**: `qr = QuantumRegister(n, 'q'); cr = ClassicalRegister(n, 'c'); qc = QuantumCircuit(qr, cr)`
- **Style C (Gate Library Append)**: `from qiskit.circuit.library import HGate; qc.append(HGate(), [0])`

> *Teleportation uses 3 qubits and 2 (or 3) classical bits. You can use standard Qiskit dynamic circuit syntax or deferred measurement.*

Write your solution in the code cell below where indicated by `# >>> YOUR CODE HERE`.

```python
# ==============================================================================
# 🛠️ STEP 3: YOUR CIRCUIT IMPLEMENTATION — Quantum State Teleportation Protocol
# ==============================================================================

# >>> YOUR CODE HERE: Step A - Circuit Initialization
# You can name your circuit 'my_circuit', 'qc', or 'circuit'
my_circuit = QuantumCircuit(3, 3)

# >>> YOUR CODE HERE: Step B - Apply Quantum Gates
# Qubit 0: Unknown state to teleport (e.g. prepared with an H gate)
# Qubits 1 & 2: Pre-shared Bell pair
my_circuit.h(0)      # State to teleport
my_circuit.h(1)      # Create Bell pair on q1, q2
my_circuit.cx(1, 2)

# Alice's Bell measurement on q0, q1
my_circuit.cx(0, 1)
my_circuit.h(0)
my_circuit.measure([0, 1], [0, 1])

# Bob's conditional corrections on q2
my_circuit.measure(2, 2)

# >>> YOUR CODE HERE: Step C - Execution & Inspection
print('--- Assembled Quantum Circuit ---')
print(my_circuit.draw(output='text'))

# Run 1,024 shots on simulator
job = sim.run(transpile(my_circuit, sim), shots=1024)
counts = job.result().get_counts()
print('\nMeasured Counts:', counts)
```
