### 📖 What You Learned in the Platform Lesson
In Chapter 4, you explored quantum entanglement—what Einstein called 'spooky action at a distance'. The canonical Bell state $|\Phi^+\rangle = \frac{|00\rangle + |11\rangle}{\sqrt{2}}$ cannot be factored into independent single-qubit states $|\psi_1\rangle \otimes |\psi_2\rangle$. Measuring one qubit instantaneously dictates the state of the other!

**Mathematical Formulation:**
$$|\Phi^+\rangle = \frac{|00\rangle + |11\rangle}{\sqrt{2}},   |\Psi^+\rangle = \frac{|01\rangle + |10\rangle}{\sqrt{2}}$$
*The canonical Bell basis states in computational representation.*

```python
# Install modern Qiskit stack (quiet mode)
!pip install -q qiskit qiskit-aer matplotlib pylatexenc numpy

import qiskit
from qiskit import QuantumCircuit, QuantumRegister, ClassicalRegister, transpile
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram
import numpy as np

# Initialize quantum simulator
sim = AerSimulator()
print(f"✅ Qiskit Core Version: {qiskit.__version__}")
print(f"✅ Aer Quantum Simulator: {sim.name}")
print("🚀 Quantum Environment Ready for Experimentation!")
```

### 🧪 Step 2: Interactive Concept Walkthrough (Reference Demonstration)
Inspect and run the reference code below. Notice how each line executes a physical transformation in Hilbert space.

```python
# Reference Implementation for: Entanglement & The 4 Bell States
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator

qc = QuantumCircuit(2, 2)
qc.h(0)
qc.cx(0, 1)
qc.measure([0, 1], [0, 1])

sim = AerSimulator()
counts = sim.run(transpile(qc, sim), shots=2000).result().get_counts()
print('Bell State Counts:', counts)

# Display circuit structure:
try:
    print('\n--- Circuit Diagram ---')
    print(qc.draw('text'))
except Exception as e:
    pass
```

#### 💡 Supported Writing Styles & Innovation
In quantum programming, there is never only one way to write a circuit! Qiskit supports multiple styles:
- **Style A (Standard Shorthand)**: `qc = QuantumCircuit(2, 2); qc.h(0)`

- **Style C (Gate Library Append)**: `from qiskit.circuit.library import HGate; qc.append(HGate(), [0])`
