### Platform Curriculum Review & Theoretical Foundations
Superdense coding is the dual counterpart of quantum teleportation. While teleportation sends 1 quantum state using 2 classical bits, superdense coding transmits 2 classical bits of information using only 1 transmitted physical qubit across a shared Bell state!

**Mathematical Formulation:**
$$\langle CHSH \rangle_{\text{quantum}} = 2\sqrt{2} \approx 2.828 > 2$$
*Tsirelson's bound violating classical Bell local realism.*

### Section 2: Reference Implementation & Theoretical Walkthrough
Analyze and execute the verified baseline implementation below. Observe how each operation transforms the state vector in Hilbert space.

```python
# Reference Implementation: Superdense Coding & The CHSH Game
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator

qc = QuantumCircuit(2, 2)
qc.h(0); qc.cx(0, 1)  # Shared EPR pair
qc.x(0); qc.z(0)      # Alice encodes '11'
qc.cx(0, 1); qc.h(0)  # Bob decodes
qc.measure([0, 1], [0, 1])

sim = AerSimulator()
counts = sim.run(transpile(qc, sim), shots=1024).result().get_counts()
print('Decoded message counts:', counts)

# Display circuit structure:
try:
    print('\n--- Circuit Structure ---')
    print(qc.draw('text'))
except Exception as e:
    pass
```

```python
# ==============================================================================
# SECTION 5: EXPERIMENTAL INVESTIGATION SANDBOX
# ==============================================================================
def encode_and_decode(message):
    qc = QuantumCircuit(2, 2)
    qc.h(0); qc.cx(0, 1) # Entangled pair
    if message[1] == '1': qc.x(0)
    if message[0] == '1': qc.z(0)
    qc.cx(0, 1); qc.h(0) # Bell measurement
    qc.measure([0, 1], [0, 1])
    return sim.run(transpile(qc, sim), shots=256).result().get_counts()

print("Transmitting '10' -> Decoded:", encode_and_decode('10'))
```
