### 🚀 Step 1: Environment Setup & Verifications
Run the cell below to install and verify **Qiskit 1.0+** and the **Qiskit Aer** simulator backend.

- `qiskit_aer.AerSimulator`: The high-performance C++ simulator running statevector and shot evolutions.

```python
# Install modern Qiskit stack (quiet mode)
!pip install -q qiskit qiskit-aer matplotlib pylatexenc numpy

import qiskit
from qiskit import QuantumCircuit, QuantumRegister, ClassicalRegister, transpile
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram
import numpy as np

# Initialize quantum simulator
sim = AerSimulator()
print(f"✅ Qiskit Core Version: {qiskit.__version__}")
print(f"✅ Aer Quantum Simulator: {sim.name}")
print("🚀 Quantum Environment Ready for Experimentation!")
```

---
### 🧪 Step 2: Interactive Concept Walkthrough (Reference Demonstration)
Inspect and run the reference code below. Notice how each line executes a physical transformation in Hilbert space.

```python
# Reference Implementation for: NISQ Constraints, Decoherence (T1, T2) & Noise Channels
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator
from qiskit_aer.noise import NoiseModel, depolarizing_error

qc = QuantumCircuit(1, 1)
qc.x(0)
qc.measure(0, 0)

nm = NoiseModel()
nm.add_all_qubit_quantum_error(depolarizing_error(0.05, 1), ['x'])

sim = AerSimulator(noise_model=nm)
counts = sim.run(transpile(qc, sim), shots=2000).result().get_counts()
print('Noisy execution counts (leakage into 0):', counts)

# Display circuit structure:
try:
    print('\n--- Circuit Diagram ---')
    print(qc.draw('text'))
except Exception as e:
    pass
```
