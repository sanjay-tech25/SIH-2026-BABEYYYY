### Platform Curriculum Review & Theoretical Foundations
In the platform lesson, you explored how quantum states exist as normalized complex vectors $|\psi\rangle \in \mathbb{C}^2$. In classical computers, numbers are binary bits (0 or 1). In Qiskit, we initialize quantum statevectors and observe how projective measurement collapses amplitudes into classical bits with probability $P(i) = |\alpha_i|^2$.

**Mathematical Formulation:**
$$⟨φ|ψ⟩ = ∑ᵢ φᵢ* ψᵢ,   with ⟨ψ|ψ⟩ = 1$$
*Inner product and normalization condition in complex Hilbert space.*
