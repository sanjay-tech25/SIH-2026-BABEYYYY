### 📖 What You Learned in the Platform Lesson
The Shor 9-Qubit Code is the landmark code demonstrating that both bit-flip errors ($X$) and phase-flip errors ($Z$) can be simultaneously corrected. It concatenates the 3-qubit phase code with the 3-qubit bit code, laying the intellectual foundation for modern 2D Surface Codes and Fault-Tolerant Quantum Computing.

**Mathematical Formulation:**
$$P_\text{logical} \propto \left(\frac{p}{p_\text{th}}\right)^{(d+1)/2},   p_\text{th} \approx 1\%$$
*Surface code exponential error suppression below physical fault-tolerance threshold.*

### 🧪 Step 2: Interactive Concept Walkthrough (Reference Demonstration)
Inspect and run the reference code below. Notice how each line executes a physical transformation in Hilbert space.

```python
# Reference Implementation for: The Shor 9-Qubit Code & Surface Codes
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator

qc = QuantumCircuit(3, 1)
qc.h(0); qc.cx(0, 1)  # Even parity Bell state
qc.cx(0, 2); qc.cx(1, 2)  # Syndrome extraction onto wire 2
qc.measure(2, 0)

sim = AerSimulator()
counts = sim.run(transpile(qc, sim), shots=1024).result().get_counts()
print('Parity syndrome counts (100% even parity 0):', counts)

# Display circuit structure:
try:
    print('\n--- Circuit Diagram ---')
    print(qc.draw('text'))
except Exception as e:
    pass
```

#### 💡 Supported Writing Styles & Innovation
In quantum programming, there is never only one way to write a circuit! Qiskit supports multiple styles:
- **Style A (Standard Shorthand)**: `qc = QuantumCircuit(9, 3); qc.h(0)`
- **Style B (Named Registers)**: `qr = QuantumRegister(n, 'q'); cr = ClassicalRegister(n, 'c'); qc = QuantumCircuit(qr, cr)`
- **Style C (Gate Library Append)**: `from qiskit.circuit.library import HGate; qc.append(HGate(), [0])`

> *Shor code uses 9 physical qubits to protect 1 logical qubit against arbitrary single-qubit errors.*

Write your solution in the code cell below where indicated by `# >>> YOUR CODE HERE`.

```python
# ==============================================================================
# 🛠️ STEP 3: YOUR CIRCUIT IMPLEMENTATION — The Shor 9-Qubit Code & Surface Codes
# ==============================================================================

# >>> YOUR CODE HERE: Step A - Circuit Initialization
# You can name your circuit 'my_circuit', 'qc', or 'circuit'
my_circuit = QuantumCircuit(9, 3)

# >>> YOUR CODE HERE: Step B - Apply Quantum Gates
# Encode logical qubit into Shor 9-qubit code
my_circuit.cx(0, 3); my_circuit.cx(0, 6)
my_circuit.h([0, 3, 6])
my_circuit.cx(0, 1); my_circuit.cx(0, 2)
my_circuit.cx(3, 4); my_circuit.cx(3, 5)
my_circuit.cx(6, 7); my_circuit.cx(6, 8)
my_circuit.measure([0, 1, 2], [0, 1, 2])

# >>> YOUR CODE HERE: Step C - Execution & Inspection
print('--- Assembled Quantum Circuit ---')
print(my_circuit.draw(output='text'))

# Run 1,024 shots on simulator
job = sim.run(transpile(my_circuit, sim), shots=1024)
counts = job.result().get_counts()
print('\nMeasured Counts:', counts)
```
