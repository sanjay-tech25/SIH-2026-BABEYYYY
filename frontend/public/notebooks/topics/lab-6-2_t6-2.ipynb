### Platform Curriculum Review & Theoretical Foundations
Grover's algorithm provides quadratic speedup $\mathcal{O}(\sqrt{N})$ for searching unstructured databases of size $N = 2^n$. It rotates the state vector in a 2D subspace spanned by the target state $|\omega\rangle$ and the uniform superposition $|s\rangle$ using alternating Oracle reflections and Diffusion operators.

**Mathematical Formulation:**
$$G = (2|s\rangle\langle s| - I) U_w,   R \approx \frac{\pi}{4}\sqrt{N}$$
*Grover diffusion operator and optimal iteration count.*

### Section 2: Reference Implementation & Theoretical Walkthrough
Analyze and execute the verified baseline implementation below. Observe how each operation transforms the state vector in Hilbert space.

```python
# Reference Implementation: Grover's Search Algorithm & Amplitude Amplification
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator

qc = QuantumCircuit(2, 2)
qc.h([0, 1])
qc.cz(0, 1)  # Oracle for |11>
qc.h([0, 1]); qc.x([0, 1]); qc.cz(0, 1); qc.x([0, 1]); qc.h([0, 1]) # Diffusion
qc.measure([0, 1], [0, 1])

sim = AerSimulator()
counts = sim.run(transpile(qc, sim), shots=1024).result().get_counts()
print('Grover search output:', counts)

# Display circuit structure:
try:
    print('\n--- Circuit Structure ---')
    print(qc.draw('text'))
except Exception as e:
    pass
```

### Section 3: Guided Hands-On Implementation
**Target Outcome**: `Measure target state |11⟩ with ~100% amplified probability`

**Implementation Directives:**
1. Prepare uniform superposition with H on wires 0 and 1.
2. Apply the Phase Oracle: CZ(0, 1) marks |11⟩.
3. Apply the Grover Diffusion Operator (H-X-CZ-X-H).
4. Measure both wires to observe ~100% counts on |11⟩.

#### Permitted Implementation Styles
In quantum programming, problems can be modeled via multiple standard paradigms. Any valid Qiskit construction is recognized:
- **Style A (Direct Shorthand)**: `qc = QuantumCircuit(2, 2); qc.h(0)`
- **Style B (Explicit Registers)**: `qr = QuantumRegister(n, 'q'); cr = ClassicalRegister(n, 'c'); qc = QuantumCircuit(qr, cr)`
- **Style C (Gate Library Appends)**: `from qiskit.circuit.library import HGate; qc.append(HGate(), [0])`

> *Note: For $n=2$, target $|11\rangle$ requires a Controlled-Z oracle `qc.cz(0, 1)` and a standard 2-qubit diffusion operator $H^{\otimes 2} X^{\otimes 2} CZ X^{\otimes 2} H^{\otimes 2}$.*

Complete your implementation in the code cell below.

```python
# ==============================================================================
# SECTION 3: CIRCUIT CONSTRUCTION — Grover's Search Algorithm & Amplitude Amplification
# ==============================================================================

# [TODO] Part A: Register Initialization
# Variable name can be 'my_circuit', 'qc', or 'circuit'
my_circuit = QuantumCircuit(2, 2)

# [TODO] Part B: Unitary Gate Application
# 1. Uniform superposition
my_circuit.h([0, 1])

# 2. Oracle marking state |11> (Controlled-Z)
my_circuit.cz(0, 1)

# 3. Grover Diffusion Operator
my_circuit.h([0, 1])
my_circuit.x([0, 1])
my_circuit.cz(0, 1)
my_circuit.x([0, 1])
my_circuit.h([0, 1])

my_circuit.measure([0, 1], [0, 1])

# [TODO] Part C: Measurement & Output Validation
print('--- Assembled Quantum Circuit ---')
print(my_circuit.draw(output='text'))

# Execute 1,024 shots on simulator
job = sim.run(transpile(my_circuit, sim), shots=1024)
counts = job.result().get_counts()
print('\nMeasured Counts:', counts)
```
