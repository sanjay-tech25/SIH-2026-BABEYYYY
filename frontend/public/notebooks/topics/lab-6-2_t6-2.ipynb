### 📖 What You Learned in the Platform Lesson
Grover's algorithm provides quadratic speedup $\mathcal{O}(\sqrt{N})$ for searching unstructured databases of size $N = 2^n$. It rotates the state vector in a 2D subspace spanned by the target state $|\omega\rangle$ and the uniform superposition $|s\rangle$ using alternating Oracle reflections and Diffusion operators.

**Mathematical Formulation:**
$$G = (2|s\rangle\langle s| - I) U_w,   R \approx \frac{\pi}{4}\sqrt{N}$$
*Grover diffusion operator and optimal iteration count.*

### 🧪 Step 2: Interactive Concept Walkthrough (Reference Demonstration)
Inspect and run the reference code below. Notice how each line executes a physical transformation in Hilbert space.

```python
# Reference Implementation for: Grover's Search Algorithm & Amplitude Amplification
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator

qc = QuantumCircuit(2, 2)
qc.h([0, 1])
qc.cz(0, 1)  # Oracle for |11>
qc.h([0, 1]); qc.x([0, 1]); qc.cz(0, 1); qc.x([0, 1]); qc.h([0, 1]) # Diffusion
qc.measure([0, 1], [0, 1])

sim = AerSimulator()
counts = sim.run(transpile(qc, sim), shots=1024).result().get_counts()
print('Grover search output:', counts)

# Display circuit structure:
try:
    print('\n--- Circuit Diagram ---')
    print(qc.draw('text'))
except Exception as e:
    pass
```

**Your Mission Goal**: Measure target state |11⟩ with ~100% amplified probability

**Step-by-Step Instructions:**

#### 💡 Supported Writing Styles & Innovation
In quantum programming, there is never only one way to write a circuit! Qiskit supports multiple styles:
- **Style A (Standard Shorthand)**: `qc = QuantumCircuit(2, 2); qc.h(0)`
- **Style B (Named Registers)**: `qr = QuantumRegister(n, 'q'); cr = ClassicalRegister(n, 'c'); qc = QuantumCircuit(qr, cr)`
- **Style C (Gate Library Append)**: `from qiskit.circuit.library import HGate; qc.append(HGate(), [0])`

> *For $n=2$, target $|11\rangle$ requires a Controlled-Z oracle `qc.cz(0, 1)` and a standard 2-qubit diffusion operator $H^{\otimes 2} X^{\otimes 2} CZ X^{\otimes 2} H^{\otimes 2}$.*

Write your solution in the code cell below where indicated by `# >>> YOUR CODE HERE`.

```python
# ==============================================================================
# 🛠️ STEP 3: YOUR CIRCUIT IMPLEMENTATION — Grover's Search Algorithm & Amplitude Amplification
# ==============================================================================

# >>> YOUR CODE HERE: Step A - Circuit Initialization
# You can name your circuit 'my_circuit', 'qc', or 'circuit'
my_circuit = QuantumCircuit(2, 2)

# >>> YOUR CODE HERE: Step B - Apply Quantum Gates
# 1. Uniform superposition
my_circuit.h([0, 1])

# 2. Oracle marking state |11> (Controlled-Z)
my_circuit.cz(0, 1)

# 3. Grover Diffusion Operator
my_circuit.h([0, 1])
my_circuit.x([0, 1])
my_circuit.cz(0, 1)
my_circuit.x([0, 1])
my_circuit.h([0, 1])

my_circuit.measure([0, 1], [0, 1])

# >>> YOUR CODE HERE: Step C - Execution & Inspection
print('--- Assembled Quantum Circuit ---')
print(my_circuit.draw(output='text'))

# Run 1,024 shots on simulator
job = sim.run(transpile(my_circuit, sim), shots=1024)
counts = job.result().get_counts()
print('\nMeasured Counts:', counts)
```
