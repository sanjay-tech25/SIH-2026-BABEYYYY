### 📖 What You Learned in the Platform Lesson
The Ekert 1991 (E91) protocol bases cryptographic security on quantum entanglement and the violation of Bell's Inequality. If an eavesdropper attempts to measure or clone the entangled particles, the Bell correlation parameter $S$ drops from $2\sqrt{2} \approx 2.828$ to below the classical threshold $S \le 2$.

**Mathematical Formulation:**
$$S = |E(a_1, b_1) - E(a_1, b_2)| + |E(a_2, b_1) + E(a_2, b_2)| = 2\sqrt{2}$$
*CHSH inequality violation certifying untrusted quantum channel security.*

### 🧪 Step 2: Interactive Concept Walkthrough (Reference Demonstration)
Inspect and run the reference code below. Notice how each line executes a physical transformation in Hilbert space.

```python
# Reference Implementation for: E91 Entanglement-Based Quantum Cryptography
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator

qc = QuantumCircuit(2, 2)
qc.h(0); qc.cx(0, 1)
qc.measure([0, 1], [0, 1])

sim = AerSimulator()
counts = sim.run(transpile(qc, sim), shots=1024).result().get_counts()
print('E91 shared key distribution:', counts)

# Display circuit structure:
try:
    print('\n--- Circuit Diagram ---')
    print(qc.draw('text'))
except Exception as e:
    pass
```
