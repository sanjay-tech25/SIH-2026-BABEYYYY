### 📖 What You Learned in the Platform Lesson
Classical repetition codes duplicate bits ($0 \to 000$). In quantum mechanics, the No-Cloning Theorem forbids copying $|\psi\rangle \to |\psi\rangle|\psi\rangle|\psi\rangle$. The 3-Qubit Bit-Flip code circumvents this by entangling 1 logical qubit into 3 physical qubits: $|\psi\rangle_L = \alpha|000\rangle + \beta|111\rangle$. Parity ancillas diagnose errors without collapsing the superposition!

**Mathematical Formulation:**
$$|0_L\rangle = |000\rangle,   |1_L\rangle = |111\rangle,   s_1 = Z_0 Z_1,   s_2 = Z_1 Z_2$$
*Logical basis encoding and stabilizer syndrome extraction operators.*

### 🧪 Step 2: Interactive Concept Walkthrough (Reference Demonstration)
Inspect and run the reference code below. Notice how each line executes a physical transformation in Hilbert space.

```python
# Reference Implementation for: The 3-Qubit Bit-Flip & Phase-Flip Codes
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator

qc = QuantumCircuit(3, 3)
qc.x(0)  # Input state |1>
qc.cx(0, 1); qc.cx(0, 2)  # Encode into |111>
qc.measure([0, 1, 2], [0, 1, 2])

sim = AerSimulator()
counts = sim.run(transpile(qc, sim), shots=1024).result().get_counts()
print('Logical qubit encoding counts:', counts)

# Display circuit structure:
try:
    print('\n--- Circuit Diagram ---')
    print(qc.draw('text'))
except Exception as e:
    pass
```
