### 📖 What You Learned in the Platform Lesson
In Chapter 6, you met the Deutsch-Jozsa algorithm—the earliest formal proof of exponential separation between quantum and classical query complexity. A classical algorithm requires up to $2^{n-1} + 1$ evaluations to guarantee whether a function is constant or balanced. The Deutsch-Jozsa algorithm decides with 100% certainty in a SINGLE quantum query!

**Mathematical Formulation:**
$$|\psi_\text{out}\rangle = \frac{1}{2^n}\sum_x \sum_y (-1)^{x \cdot y + f(x)}|y\rangle$$
*Hadamard transform interference concentrating amplitude on function property.*

### 🧪 Step 2: Interactive Concept Walkthrough (Reference Demonstration)
Inspect and run the reference code below. Notice how each line executes a physical transformation in Hilbert space.

```python
# Reference Implementation for: Deutsch-Jozsa & Bernstein-Vazirani Oracles
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator

qc = QuantumCircuit(3, 2)
qc.h([0, 1]); qc.x(2); qc.h(2)  # Superposition and ancilla |->
qc.cx(0, 2); qc.cx(1, 2)         # Oracle for hidden string '11'
qc.h([0, 1])                     # Interference
qc.measure([0, 1], [0, 1])

sim = AerSimulator()
counts = sim.run(transpile(qc, sim), shots=1024).result().get_counts()
print('Recovered hidden string:', counts)

# Display circuit structure:
try:
    print('\n--- Circuit Diagram ---')
    print(qc.draw('text'))
except Exception as e:
    pass
```

#### 💡 Supported Writing Styles & Innovation
In quantum programming, there is never only one way to write a circuit! Qiskit supports multiple styles:
- **Style A (Standard Shorthand)**: `qc = QuantumCircuit(3, 2); qc.h(0)`
- **Style B (Named Registers)**: `qr = QuantumRegister(n, 'q'); cr = ClassicalRegister(n, 'c'); qc = QuantumCircuit(qr, cr)`
- **Style C (Gate Library Append)**: `from qiskit.circuit.library import HGate; qc.append(HGate(), [0])`

> *Initialize input qubits in $|+\rangle$ and ancilla in $|-\rangle$. Evaluate oracle. Apply Hadamards to input qubits.*

Write your solution in the code cell below where indicated by `# >>> YOUR CODE HERE`.

```python
# ==============================================================================
# 🛠️ STEP 3: YOUR CIRCUIT IMPLEMENTATION — Deutsch-Jozsa & Bernstein-Vazirani Oracles
# ==============================================================================

# >>> YOUR CODE HERE: Step A - Circuit Initialization
# You can name your circuit 'my_circuit', 'qc', or 'circuit'
my_circuit = QuantumCircuit(3, 2)

# >>> YOUR CODE HERE: Step B - Apply Quantum Gates
# 2 input qubits (q0, q1) in |+>, 1 ancilla (q2) in |->
my_circuit.h(0)
my_circuit.h(1)
my_circuit.x(2); my_circuit.h(2)

# Balanced Oracle: flip ancilla if q0 == 1 (CX from q0 to q2)
my_circuit.cx(0, 2)

# Final Hadamards on inputs
my_circuit.h(0)
my_circuit.h(1)
my_circuit.measure([0, 1], [0, 1])

# >>> YOUR CODE HERE: Step C - Execution & Inspection
print('--- Assembled Quantum Circuit ---')
print(my_circuit.draw(output='text'))

# Run 1,024 shots on simulator
job = sim.run(transpile(my_circuit, sim), shots=1024)
counts = job.result().get_counts()
print('\nMeasured Counts:', counts)
```
