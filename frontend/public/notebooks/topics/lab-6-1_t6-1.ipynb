### Platform Curriculum Review & Theoretical Foundations
In Chapter 6, you met the Deutsch-Jozsa algorithm—the earliest formal proof of exponential separation between quantum and classical query complexity. A classical algorithm requires up to $2^{n-1} + 1$ evaluations to guarantee whether a function is constant or balanced. The Deutsch-Jozsa algorithm decides with 100% certainty in a SINGLE quantum query!

**Mathematical Formulation:**
$$|\psi_\text{out}\rangle = \frac{1}{2^n}\sum_x \sum_y (-1)^{x \cdot y + f(x)}|y\rangle$$
*Hadamard transform interference concentrating amplitude on function property.*

### Section 2: Reference Implementation & Theoretical Walkthrough
Analyze and execute the verified baseline implementation below. Observe how each operation transforms the state vector in Hilbert space.

```python
# Reference Implementation: Deutsch-Jozsa & Bernstein-Vazirani Oracles
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator

qc = QuantumCircuit(3, 2)
qc.h([0, 1]); qc.x(2); qc.h(2)  # Superposition and ancilla |->
qc.cx(0, 2); qc.cx(1, 2)         # Oracle for hidden string '11'
qc.h([0, 1])                     # Interference
qc.measure([0, 1], [0, 1])

sim = AerSimulator()
counts = sim.run(transpile(qc, sim), shots=1024).result().get_counts()
print('Recovered hidden string:', counts)

# Display circuit structure:
try:
    print('\n--- Circuit Structure ---')
    print(qc.draw('text'))
except Exception as e:
    pass
```

#### Permitted Implementation Styles
In quantum programming, problems can be modeled via multiple standard paradigms. Any valid Qiskit construction is recognized:
- **Style A (Direct Shorthand)**: `qc = QuantumCircuit(3, 2); qc.h(0)`
- **Style B (Explicit Registers)**: `qr = QuantumRegister(n, 'q'); cr = ClassicalRegister(n, 'c'); qc = QuantumCircuit(qr, cr)`
- **Style C (Gate Library Appends)**: `from qiskit.circuit.library import HGate; qc.append(HGate(), [0])`

```python
# ==============================================================================
# SECTION 3: CIRCUIT CONSTRUCTION — Deutsch-Jozsa & Bernstein-Vazirani Oracles
# ==============================================================================

# [TODO] Part A: Register Initialization
# Variable name can be 'my_circuit', 'qc', or 'circuit'
my_circuit = QuantumCircuit(3, 2)

# [TODO] Part B: Unitary Gate Application
# 2 input qubits (q0, q1) in |+>, 1 ancilla (q2) in |->
my_circuit.h(0)
my_circuit.h(1)
my_circuit.x(2); my_circuit.h(2)

# Balanced Oracle: flip ancilla if q0 == 1
my_circuit.cx(0, 2)

# Final Hadamards on inputs
my_circuit.h(0)
my_circuit.h(1)
my_circuit.measure([0, 1], [0, 1])

# [TODO] Part C: Measurement & Output Validation
print('--- Assembled Quantum Circuit ---')
print(my_circuit.draw(output='text'))

# Execute 1,024 shots on simulator
job = sim.run(transpile(my_circuit, sim), shots=1024)
counts = job.result().get_counts()
print('\nMeasured Counts:', counts)
```
