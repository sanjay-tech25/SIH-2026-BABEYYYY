### Platform Curriculum Review & Theoretical Foundations
In Qiskit 1.0+, execution is standardized around Execution Primitives: SamplerV2 evaluates probability distributions from measurement bitstrings, and EstimatorV2 computes physical expectation values $\langle O \rangle = \langle\psi|O|\psi\rangle$ with automated error mitigation. This mirrors how commercial quantum computers are programmed today.

**Mathematical Formulation:**
$$\text{EstimatorV2}: \langle O \rangle = \text{Tr}(\rho O),   \text{SamplerV2}: P(x) = |\langle x|\psi\rangle|^2$$
*Mathematical definitions of Qiskit Runtime Primitives Estimator and Sampler.*

```python
# Install modern Qiskit stack (quiet mode)
!pip install -q qiskit qiskit-aer matplotlib pylatexenc numpy

import qiskit
from qiskit import QuantumCircuit, QuantumRegister, ClassicalRegister, transpile
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram
import numpy as np

# Initialize simulator backend
sim = AerSimulator()
print(f"[STATUS] Qiskit Core Version: {qiskit.__version__}")
print(f"[STATUS] Aer Simulator Backend: {sim.name}")
print("[OK] Quantum execution environment ready.")
```

### Section 2: Reference Implementation & Theoretical Walkthrough
Analyze and execute the verified baseline implementation below. Observe how each operation transforms the state vector in Hilbert space.

```python
# Reference Implementation: Quantum Runtime Primitives: SamplerV2 & EstimatorV2
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator

qc = QuantumCircuit(2, 2)
qc.h(0); qc.cx(0, 1)
qc.measure([0, 1], [0, 1])

sim = AerSimulator()
job = sim.run(transpile(qc, sim), shots=1024)
counts = job.result().get_counts()
print('Runtime Sampler output:', counts)

# Display circuit structure:
try:
    print('\n--- Circuit Structure ---')
    print(qc.draw('text'))
except Exception as e:
    pass
```

```python
# ==============================================================================
# SECTION 5: EXPERIMENTAL INVESTIGATION SANDBOX
# ==============================================================================
from qiskit.quantum_info import Statevector, SparsePauliOp

bell = QuantumCircuit(2)
bell.h(0); bell.cx(0, 1)
sv = Statevector.from_instruction(bell)
observable = SparsePauliOp(["ZZ", "XX"])
exp_val = sv.expectation_value(observable)
print("Observable expectation value <ZZ + XX>:", exp_val.real)
```
