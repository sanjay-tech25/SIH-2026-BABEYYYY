In Qiskit 1.0+, execution is standardized around Execution Primitives: SamplerV2 evaluates probability distributions from measurement bitstrings, and EstimatorV2 computes physical expectation values $\langle O \rangle = \langle\psi|O|\psi\rangle$ with automated error mitigation. This mirrors how commercial quantum computers are programmed today.

**Mathematical Formulation:**
$$\text{EstimatorV2}: \langle O \rangle = \text{Tr}(\rho O),   \text{SamplerV2}: P(x) = |\langle x|\psi\rangle|^2$$
*Mathematical definitions of Qiskit Runtime Primitives Estimator and Sampler.*

### 🚀 Step 1: Environment Setup & Verifications
Run the cell below to install and verify **Qiskit 1.0+** and the **Qiskit Aer** simulator backend.
- `qiskit.QuantumCircuit`: The central container for registers and gate transformations.
- `qiskit_aer.AerSimulator`: The high-performance C++ simulator running statevector and shot evolutions.
- `qiskit.visualization`: Tools to render circuit diagrams and probability histograms.

```python
# Install modern Qiskit stack (quiet mode)
!pip install -q qiskit qiskit-aer matplotlib pylatexenc numpy

import qiskit
from qiskit import QuantumCircuit, QuantumRegister, ClassicalRegister, transpile
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram
import numpy as np

# Initialize quantum simulator
sim = AerSimulator()
print(f"✅ Qiskit Core Version: {qiskit.__version__}")
print(f"✅ Aer Quantum Simulator: {sim.name}")
print("🚀 Quantum Environment Ready for Experimentation!")
```

### 🧪 Step 2: Interactive Concept Walkthrough (Reference Demonstration)
Inspect and run the reference code below. Notice how each line executes a physical transformation in Hilbert space.

```python
# Reference Implementation for: Quantum Runtime Primitives: SamplerV2 & EstimatorV2
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator

qc = QuantumCircuit(2, 2)
qc.h(0); qc.cx(0, 1)
qc.measure([0, 1], [0, 1])

sim = AerSimulator()
job = sim.run(transpile(qc, sim), shots=1024)
counts = job.result().get_counts()
print('Runtime Sampler output:', counts)

# Display circuit structure:
try:
    print('\n--- Circuit Diagram ---')
    print(qc.draw('text'))
except Exception as e:
    pass
```

```python
# ==============================================================================
# 🚀 STEP 5: YOUR INNOVATION & CREATIVE EXPERIMENTATION SANDBOX
# ==============================================================================
from qiskit.quantum_info import Statevector, SparsePauliOp

bell = QuantumCircuit(2)
bell.h(0); bell.cx(0, 1)
sv = Statevector.from_instruction(bell)
observable = SparsePauliOp(["ZZ", "XX"])
exp_val = sv.expectation_value(observable)
print("Expectation value of ZZ + XX on |Phi+>:", exp_val.real)
```
