### 📖 What You Learned in the Platform Lesson
Quantum Phase Estimation (QPE) is the engine behind Shor's factoring algorithm and quantum chemistry simulation. Given a unitary operator $U$ with eigenstate $|u\rangle$ such that $U|u\rangle = e^{2\pi i \theta}|u\rangle$, QPE estimates the phase $\theta \in [0, 1)$ with exponential precision using inverse Quantum Fourier Transform.

**Mathematical Formulation:**
$$U|u\rangle = e^{2\pi i \theta}|u\rangle,   \mathcal{O}((\log N)^3)$$
*Eigenvalue phase estimation and Shor's cubic polynomial complexity.*

#### 💡 Supported Writing Styles & Innovation
In quantum programming, there is never only one way to write a circuit! Qiskit supports multiple styles:
- **Style A (Standard Shorthand)**: `qc = QuantumCircuit(2, 1); qc.h(0)`
- **Style B (Named Registers)**: `qr = QuantumRegister(n, 'q'); cr = ClassicalRegister(n, 'c'); qc = QuantumCircuit(qr, cr)`
- **Style C (Gate Library Append)**: `from qiskit.circuit.library import HGate; qc.append(HGate(), [0])`

> *Use counting qubits and target qubit. Apply controlled unitaries $U^{2^j}$ and inverse QFT.*

Write your solution in the code cell below where indicated by `# >>> YOUR CODE HERE`.

```python
# ==============================================================================
# 🛠️ STEP 3: YOUR CIRCUIT IMPLEMENTATION — Quantum Phase Estimation & Shor's Algorithm
# ==============================================================================

# >>> YOUR CODE HERE: Step A - Circuit Initialization
# You can name your circuit 'my_circuit', 'qc', or 'circuit'
my_circuit = QuantumCircuit(2, 1)

# >>> YOUR CODE HERE: Step B - Apply Quantum Gates
# Estimating phase of S gate (theta = 0.25 -> binary 0.01)
my_circuit.x(1)       # Prepare eigenstate |1>
my_circuit.h(0)       # Counting qubit in |+>
my_circuit.cs(0, 1)   # Controlled-S gate
my_circuit.h(0)       # 1-qubit inverse QFT
my_circuit.measure(0, 0)

# >>> YOUR CODE HERE: Step C - Execution & Inspection
print('--- Assembled Quantum Circuit ---')
print(my_circuit.draw(output='text'))

# Run 1,024 shots on simulator
job = sim.run(transpile(my_circuit, sim), shots=1024)
counts = job.result().get_counts()
print('\nMeasured Counts:', counts)
```
