### 📖 What You Learned in the Platform Lesson
The BB84 protocol (Bennett & Brassard 1984) guarantees unconditionally secure communication rooted in the laws of quantum mechanics. Eavesdropping inevitably disturbs quantum states due to the No-Cloning Theorem, introducing detectable errors ($QBER > 11\%$) during basis sifting.

**Mathematical Formulation:**
$$\text{QBER} = \frac{E_\text{error}}{N_\text{sifted}},   \text{Eve introduces } \ge 25\% \text{ QBER}$$
*Quantum Bit Error Rate threshold guaranteeing eavesdropper detection.*

```python
# Install modern Qiskit stack (quiet mode)
!pip install -q qiskit qiskit-aer matplotlib pylatexenc numpy

import qiskit
from qiskit import QuantumCircuit, QuantumRegister, ClassicalRegister, transpile
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram
import numpy as np

# Initialize quantum simulator
sim = AerSimulator()
print(f"✅ Qiskit Core Version: {qiskit.__version__}")
print(f"✅ Aer Quantum Simulator: {sim.name}")
print("🚀 Quantum Environment Ready for Experimentation!")
```

### 🧪 Step 2: Interactive Concept Walkthrough (Reference Demonstration)
Inspect and run the reference code below. Notice how each line executes a physical transformation in Hilbert space.

```python
# Reference Implementation for: The BB84 Protocol & Quantum Key Distribution
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator

qc = QuantumCircuit(1, 1)
qc.h(0)  # Alice encodes in X-basis
qc.h(0)  # Bob measures in matching X-basis
qc.measure(0, 0)

sim = AerSimulator()
counts = sim.run(transpile(qc, sim), shots=1024).result().get_counts()
print('BB84 matching basis counts (100% 0):', counts)

# Display circuit structure:
try:
    print('\n--- Circuit Diagram ---')
    print(qc.draw('text'))
except Exception as e:
    pass
```
