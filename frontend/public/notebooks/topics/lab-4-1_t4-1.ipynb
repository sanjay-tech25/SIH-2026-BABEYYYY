### Platform Curriculum Review & Theoretical Foundations
In Chapter 4, you explored quantum entanglement—what Einstein called 'spooky action at a distance'. The canonical Bell state $|\Phi^+\rangle = \frac{|00\rangle + |11\rangle}{\sqrt{2}}$ cannot be factored into independent single-qubit states $|\psi_1\rangle \otimes |\psi_2\rangle$. Measuring one qubit instantaneously dictates the state of the other!

**Mathematical Formulation:**
$$|\Phi^+\rangle = \frac{|00\rangle + |11\rangle}{\sqrt{2}},   |\Psi^+\rangle = \frac{|01\rangle + |10\rangle}{\sqrt{2}}$$
*The canonical Bell basis states in computational representation.*

### Section 2: Reference Implementation & Theoretical Walkthrough
Analyze and execute the verified baseline implementation below. Observe how each operation transforms the state vector in Hilbert space.

```python
# Reference Implementation: Entanglement & The 4 Bell States
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator

qc = QuantumCircuit(2, 2)
qc.h(0)
qc.cx(0, 1)
qc.measure([0, 1], [0, 1])

sim = AerSimulator()
counts = sim.run(transpile(qc, sim), shots=2000).result().get_counts()
print('Bell State Counts:', counts)

# Display circuit structure:
try:
    print('\n--- Circuit Structure ---')
    print(qc.draw('text'))
except Exception as e:
    pass
```

### Section 3: Guided Hands-On Implementation
**Target Outcome**: `Measure strictly correlated 50% |00⟩ and 50% |11⟩`

**Implementation Directives:**
1. Put wire 0 into superposition with an H gate.

#### Permitted Implementation Styles
In quantum programming, problems can be modeled via multiple standard paradigms. Any valid Qiskit construction is recognized:
- **Style A (Direct Shorthand)**: `qc = QuantumCircuit(2, 2); qc.h(0)`
- **Style B (Explicit Registers)**: `qr = QuantumRegister(n, 'q'); cr = ClassicalRegister(n, 'c'); qc = QuantumCircuit(qr, cr)`
- **Style C (Gate Library Appends)**: `from qiskit.circuit.library import HGate; qc.append(HGate(), [0])`

> *Note: Create $|\Phi^+\rangle$ using `qc.h(0)` followed by `qc.cx(0, 1)`. You can also create the other 3 Bell states by adding $X$ or $Z$ gates before the entangler.*

Complete your implementation in the code cell below.

```python
# ==============================================================================
# SECTION 3: CIRCUIT CONSTRUCTION — Entanglement & The 4 Bell States
# ==============================================================================

# [TODO] Part A: Register Initialization
# Variable name can be 'my_circuit', 'qc', or 'circuit'
my_circuit = QuantumCircuit(2, 2)

# [TODO] Part B: Unitary Gate Application
# 1. Put qubit 0 into superposition with Hadamard
my_circuit.h(0)
# 2. Entangle qubit 0 and qubit 1 with CNOT
my_circuit.cx(0, 1)
# 3. Measure both qubits
my_circuit.measure([0, 1], [0, 1])

# [TODO] Part C: Measurement & Output Validation
print('--- Assembled Quantum Circuit ---')
print(my_circuit.draw(output='text'))

# Execute 1,024 shots on simulator
job = sim.run(transpile(my_circuit, sim), shots=1024)
counts = job.result().get_counts()
print('\nMeasured Counts:', counts)
```
