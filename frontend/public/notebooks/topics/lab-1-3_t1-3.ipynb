### 📚 Pedagogical Background & Theory
Any single-qubit pure state can be visualized as a point on the surface of the unit Bloch sphere:

$$|\psi\rangle = \cos(\theta/2)|0\rangle + e^{i\phi}\sin(\theta/2)|1\rangle$$

- North Pole ($\theta = 0$): $|0\rangle$
- South Pole ($\theta = \pi$): $|1\rangle$
- Equator ($\theta = \pi/2$): $|+\rangle$ ($\phi=0$), $|-\rangle$ ($\phi=\pi$), $|+i\rangle$ ($\phi=\pi/2$).
The Pauli-X gate acts as a bit-flip (NOT gate), rotating $\pi$ radians around the X-axis.

### 📖 Step 2: Interactive Concept Walkthrough (Reference Demonstration)
Study this pre-built demonstration illustrating the physical principles of this topic.

```python
# Rotating from |0> to |1> via Pauli-X gate
qc_x = QuantumCircuit(1, 1)
qc_x.x(0)  # Rotates to south pole: |1>
qc_x.measure(0, 0)
print(qc_x.draw(output='text'))
print('Measured counts:', sim.run(transpile(qc_x, sim), shots=1024).result().get_counts())
```
