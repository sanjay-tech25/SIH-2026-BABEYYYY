**Pedagogical Guidance:** In the platform lesson, you explored how quantum states exist as normalized complex vectors $|\psi\rangle \in \mathbb{C}^2$. ### Topic 3.1: Pauli Operators (X, Y, Z) & Phase Shifting (S, T)
**Mission Objective:** Construct an H-S-S-H Mach-Zehnder circuit to observe destructive interference caused by a π phase shift.

**Pedagogical Guidance:** In Chapter 3, you discovered the single-qubit Pauli family: $X$ flips bits, $Z$ flips the relative phase of $|1\rangle$, $Y = iXZ$ does both, and phase gates $S = Z^{1/2}, T = Z^{1/4}$ rotate about the Z-axis by $\pi/2$ and $\pi/4$. Let's demonstrate how a phase flip $Z$ converts $|+\rangle$ into $|-\rangle$!

**Target Outcome:** `Measure 100% probability for state |1⟩ via interference`


```python
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator

qc = QuantumCircuit(1, 1)
qc.h(0)
qc.s(0)
qc.s(0)
qc.h(0)
qc.measure(0, 0)

sim = AerSimulator()
counts = sim.run(transpile(qc, sim), shots=1024).result().get_counts()
print('H-S-S-H Interference counts:', counts)
```

**Implementation Directives:**
- Place an H gate on wire 0.

**Experimental Variation:** Investigation 1: Verify algebraic identity $T^4 = Z$ using operator equivalence checks.

**Pedagogical Guidance:** Two-qubit gates are the engine of quantum advantage. The Controlled-NOT (CNOT or CX) gate entangles qubits by flipping target qubit $q_1$ if and only if control qubit $q_0 = |1\rangle$. In matrix terms, it acts as the 4x4 permutation matrix.

### Topic 4.1: Entanglement & The 4 Bell States
**Mission Objective:** Generate the maximally entangled Bell pair |Φ⁺⟩ using Hadamard and CNOT, verifying 0% leakage into |01⟩/|10⟩.

**Pedagogical Guidance:** In Chapter 4, you explored quantum entanglement—what Einstein called 'spooky action at a distance'. The canonical Bell state $|\Phi^+\rangle = \frac{|00\rangle + |11\rangle}{\sqrt{2}}$ cannot be factored into independent single-qubit states $|\psi_1\rangle \otimes |\psi_2\rangle$. Measuring one qubit instantaneously dictates the state of the other!

**Target Outcome:** `Measure strictly correlated 50% |00⟩ and 50% |11⟩`


```python
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator

qc = QuantumCircuit(2, 2)
qc.h(0)
qc.cx(0, 1)
qc.measure([0, 1], [0, 1])

sim = AerSimulator()
counts = sim.run(transpile(qc, sim), shots=2000).result().get_counts()
print('Bell State Counts:', counts)
```

**Implementation Directives:**

### Topic 4.2: Phase Kickback & Quantum Oracles
**Mission Objective:** Prepare wire 1 in state |-⟩, apply CNOT(0, 1), and observe the phase shift reflected on wire 0.

**Pedagogical Guidance:** Phase kickback is the mathematical secret weapon powering Deutsch-Jozsa, Grover, and Shor's algorithms. When a controlled gate acts on a target qubit prepared in an eigenstate with eigenvalue -1 (specifically $|-\rangle = \frac{|0\rangle - |1\rangle}{\sqrt{2}}$), the phase kicks back into the control qubit!

**Target Outcome:** `Observe Phase Kickback transforming control wire into |-⟩`


```python
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator

qc = QuantumCircuit(2, 1)
qc.h(0)     # Control in |+>
qc.x(1)     # Target to |1>
qc.h(1)     # Target to |->
qc.cx(0, 1) # Phase kickback flips wire 0 to |->
qc.h(0)     # H transforms |-> back to |1>
qc.measure(0, 0)

sim = AerSimulator()
counts = sim.run(transpile(qc, sim), shots=1024).result().get_counts()
print('Kickback result (100% outcome 1):', counts)
```

**Implementation Directives:**
- Set control wire 0 into |+⟩ using H(0).
- Set target wire 1 into |-⟩ using X(1) followed by H(1).
- Apply CNOT(0, 1) and inspect how wire 0 transforms from |+⟩ into |-⟩.

**Experimental Variation:** Investigation 1: Demonstrate phase kickback using a Controlled-Phase operator `qc.cp(np.pi/2, 0, 1)`.

Transfer unknown quantum states using classical bits and entanglement (Teleportation), send 2 classical bits using 1 qubit (Superdense Coding), and test non-locality with CHSH.


### Topic 5.1: Quantum State Teleportation Protocol
**Mission Objective:** Assemble the complete 3-qubit teleportation circuit and transmit state |1⟩ from Alice to Bob.

**Pedagogical Guidance:** Quantum teleportation is the bedrock of distributed quantum computing and the Quantum Internet. Because of the No-Cloning Theorem, unknown quantum information cannot be copied. Teleportation transfers the complete quantum state $|\psi\rangle = \alpha|0\rangle + \beta|1\rangle$ from Qubit 0 to Qubit 2 using shared entanglement and 2 classical bits.

**Target Outcome:** `Successfully teleport quantum state to wire 2`


```python
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator

qc = QuantumCircuit(3, 2)
qc.x(0)  # Teleport state |1>
qc.h(1); qc.cx(1, 2)  # Bell pair
qc.cx(0, 1); qc.h(0)  # Alice measurement
qc.measure([0, 1], [0, 1])

sim = AerSimulator()
print(qc.draw(output='text'))
```

**Implementation Directives:**
- Prepare state to teleport on wire 0 (apply X to set |1⟩).

### Topic 5.2: Superdense Coding & The CHSH Game
**Mission Objective:** Encode 2 classical bits ('11') into a shared Bell pair and decode both bits on Bob's register.

**Pedagogical Guidance:** Superdense coding is the dual counterpart of quantum teleportation. While teleportation sends 1 quantum state using 2 classical bits, superdense coding transmits 2 classical bits of information using only 1 transmitted physical qubit across a shared Bell state!

**Target Outcome:** `Decode 100% probability for classical message '11'`


```python
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator

qc = QuantumCircuit(2, 2)
qc.h(0); qc.cx(0, 1)  # Shared EPR pair
qc.x(0); qc.z(0)      # Alice encodes '11'
qc.cx(0, 1); qc.h(0)  # Bob decodes
qc.measure([0, 1], [0, 1])

sim = AerSimulator()
counts = sim.run(transpile(qc, sim), shots=1024).result().get_counts()
print('Decoded message counts:', counts)
```

**Implementation Directives:**
- Initialize shared Bell pair on wires 0 and 1.
- Alice encodes '11' by applying X and Z gates to wire 0.
- Bob decodes by applying CNOT(0, 1) and H(0), then measuring both wires.

**Experimental Variation:** Investigation 1: Test transmission of message '01' (unitary X) and confirm Bob decodes '01'.

# Chapter 6: Foundational Quantum Algorithms
**Subtitle:** Deutsch-Jozsa, Bernstein-Vazirani & Grover's Search | **Knowledge Base Domain:** `06 - Quantum Algorithms`
Implement the landmark algorithms demonstrating exponential and quadratic quantum advantage: Deutsch-Jozsa, Bernstein-Vazirani, and Grover's search.


### Topic 6.1: Deutsch-Jozsa & Bernstein-Vazirani Oracles
**Mission Objective:** Discover a hidden 2-bit string ('11') in exactly 1 quantum query using phase kickback.

**Pedagogical Guidance:** In Chapter 6, you met the Deutsch-Jozsa algorithm—the earliest formal proof of exponential separation between quantum and classical query complexity. A classical algorithm requires up to $2^{n-1} + 1$ evaluations to guarantee whether a function is constant or balanced. The Deutsch-Jozsa algorithm decides with 100% certainty in a SINGLE quantum query!

**Target Outcome:** `Measure hidden bitstring '11' with 100% probability in 1 query`


```python
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator

qc = QuantumCircuit(3, 2)
qc.h([0, 1]); qc.x(2); qc.h(2)  # Superposition and ancilla |->
qc.cx(0, 2); qc.cx(1, 2)         # Oracle for hidden string '11'
qc.h([0, 1])                     # Interference
qc.measure([0, 1], [0, 1])

sim = AerSimulator()
counts = sim.run(transpile(qc, sim), shots=1024).result().get_counts()
print('Recovered hidden string:', counts)
```

**Implementation Directives:**
- Initialize 2 data qubits and 1 ancilla in |-⟩.
- Apply Hadamard to all wires.
- Apply CNOT oracle for hidden string '11'.
- Apply closing Hadamards and measure data wires to reveal '11'.

**Experimental Variation:** Investigation 1: Construct a constant oracle (identity operation on ancilla) and confirm deterministic output |00>.


### Topic 6.2: Grover's Search Algorithm & Amplitude Amplification
**Mission Objective:** Implement the complete Grover search algorithm to amplify target item |11⟩ to 100% probability.

**Pedagogical Guidance:** Grover's algorithm provides quadratic speedup $\mathcal{O}(\sqrt{N})$ for searching unstructured databases of size $N = 2^n$. It rotates the state vector in a 2D subspace spanned by the target state $|\omega\rangle$ and the uniform superposition $|s\rangle$ using alternating Oracle reflections and Diffusion operators.

**Target Outcome:** `Measure target state |11⟩ with ~100% amplified probability`


```python
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator

qc = QuantumCircuit(2, 2)
qc.h([0, 1])
qc.cz(0, 1)  # Oracle for |11>
qc.h([0, 1]); qc.x([0, 1]); qc.cz(0, 1); qc.x([0, 1]); qc.h([0, 1]) # Diffusion
qc.measure([0, 1], [0, 1])

sim = AerSimulator()
counts = sim.run(transpile(qc, sim), shots=1024).result().get_counts()
print('Grover search output:', counts)
```

**Implementation Directives:**
- Prepare uniform superposition with H on wires 0 and 1.
- Apply the Phase Oracle: CZ(0, 1) marks |11⟩.
- Apply the Grover Diffusion Operator (H-X-CZ-X-H).
- Measure both wires to observe ~100% counts on |11⟩.

**Experimental Variation:** Investigation 1: Reconfigure the oracle to mark state $|01\rangle$ by conjugating qubit 1 with X gates.

### Topic 7.1: Quantum Phase Estimation & Shor's Algorithm
**Mission Objective:** Build a 2-qubit Quantum Fourier Transform circuit using Hadamard and Controlled-Phase rotations.

**Pedagogical Guidance:** Quantum Phase Estimation (QPE) is the engine behind Shor's factoring algorithm and quantum chemistry simulation. Given a unitary operator $U$ with eigenstate $|u\rangle$ such that $U|u\rangle = e^{2\pi i \theta}|u\rangle$, QPE estimates the phase $\theta \in [0, 1)$ with exponential precision using inverse Quantum Fourier Transform.

**Target Outcome:** `Synthesize discrete quantum Fourier transform state`


```python
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator
import numpy as np

qc = QuantumCircuit(2)
qc.h(0)
qc.cp(np.pi / 2, 1, 0)
qc.h(1)
qc.swap(0, 1)
print(qc.draw(output='text'))
```

**Implementation Directives:**
- Apply H to wire 0.
- Apply Controlled-Phase CPhase(π/2) from wire 1 to wire 0.
- Apply H to wire 1 and insert a final SWAP gate.

**Experimental Variation:** Investigation 1: Estimate phase of the T gate (\theta = 0.125) using a 3-qubit counting register.


### Topic 7.2: Variational Quantum Eigensolver (VQE) & QAOA
**Mission Objective:** Construct a parameterized Ry rotation ansätz and sweep parameter θ from 0 to π to find the energy minimum.

**Pedagogical Guidance:** Variational Quantum Eigensolver (VQE) is the leading NISQ algorithm for quantum chemistry and materials science. It evaluates the ground state energy of a molecular Hamiltonian $H$ using a parameterized trial state $|\psi(\vec{\theta})\rangle$, minimizing $E(\vec{\theta}) = \langle\psi(\vec{\theta})|H|\psi(\vec{\theta})\rangle$ via classical optimization.

**Target Outcome:** `Measure 100% |1⟩ at bound parameter θ = π`


```python
from qiskit import QuantumCircuit, transpile
from qiskit.circuit import Parameter
from qiskit_aer import AerSimulator
import numpy as np

theta = Parameter('θ')
qc = QuantumCircuit(1, 1)
qc.ry(theta, 0)
qc.measure(0, 0)

sim = AerSimulator()
bound = qc.assign_parameters({theta: np.pi})
counts = sim.run(transpile(bound, sim), shots=1024).result().get_counts()
print('VQE ground state counts:', counts)
```

**Implementation Directives:**
- Construct a 1-qubit circuit with parameterized Ry(θ) gate.
- Bind parameter θ = π to rotate state to |1⟩ (eigenstate of Pauli-Z with energy -1).
- Measure in Z-basis to confirm 100% probability for |1⟩.

**Experimental Variation:** Investigation 1: Sweep \theta \in [0, 2\pi] across 20 parameters and reconstruct the energy surface.

### Topic 9.1: The 3-Qubit Bit-Flip & Phase-Flip Codes
**Mission Objective:** Encode logical state |1_L⟩ = |111⟩ and verify that single-qubit bit flips can be detected via parity checks.

**Pedagogical Guidance:** Classical repetition codes duplicate bits ($0 \to 000$). In quantum mechanics, the No-Cloning Theorem forbids copying $|\psi\rangle \to |\psi\rangle|\psi\rangle|\psi\rangle$. The 3-Qubit Bit-Flip code circumvents this by entangling 1 logical qubit into 3 physical qubits: $|\psi\rangle_L = \alpha|000\rangle + \beta|111\rangle$. Parity ancillas diagnose errors without collapsing the superposition!

**Target Outcome:** `Measure 100% |111⟩ for logical state |1_L⟩`


```python
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator

qc = QuantumCircuit(3, 3)
qc.x(0)  # Input state |1>
qc.cx(0, 1); qc.cx(0, 2)  # Encode into |111>
qc.measure([0, 1, 2], [0, 1, 2])

sim = AerSimulator()
counts = sim.run(transpile(qc, sim), shots=1024).result().get_counts()
print('Logical qubit encoding counts:', counts)
```

**Implementation Directives:**
- Set data qubit 0 to |1⟩ with an X gate.
- Encode into 3 physical qubits using CNOT(0, 1) and CNOT(0, 2).
- Measure all 3 wires to verify 100% collapse into |111⟩.

**Experimental Variation:** Investigation 1: Construct the 3-Qubit Phase-Flip code by wrapping the physical code in Hadamard gates.


### Topic 9.2: The Shor 9-Qubit Code & Surface Codes
**Mission Objective:** Measure two-qubit parity Z_0 Z_1 using an ancilla qubit without collapsing the logical superposition.

**Pedagogical Guidance:** The Shor 9-Qubit Code is the landmark code demonstrating that both bit-flip errors ($X$) and phase-flip errors ($Z$) can be simultaneously corrected. It concatenates the 3-qubit phase code with the 3-qubit bit code, laying the intellectual foundation for modern 2D Surface Codes and Fault-Tolerant Quantum Computing.

**Target Outcome:** `Extract even parity syndrome 0 on ancilla wire`

### Topic 10.1: NISQ Constraints, Decoherence (T1, T2) & Noise Channels
**Mission Objective:** Simulate a 5% depolarizing noise channel and observe state degradation and thermal leakage into state |0⟩.

**Pedagogical Guidance:** Real quantum processors (NISQ era) operate in noisy thermal environments. Decoherence channels degrade pure quantum states into mixed states via energy relaxation ($T_1$) and dephasing ($T_2$). In Qiskit Aer, we simulate these physical channels using `NoiseModel` and `thermal_relaxation_error`.

**Target Outcome:** `Observe realistic NISQ noise leakage into state |0⟩`


```python
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator
from qiskit_aer.noise import NoiseModel, depolarizing_error

qc = QuantumCircuit(1, 1)
qc.x(0)
qc.measure(0, 0)

nm = NoiseModel()
nm.add_all_qubit_quantum_error(depolarizing_error(0.05, 1), ['x'])

sim = AerSimulator(noise_model=nm)
counts = sim.run(transpile(qc, sim), shots=2000).result().get_counts()
print('Noisy execution counts (leakage into 0):', counts)
```

### Topic 10.2: Zero-Noise Extrapolation (ZNE) & Error Mitigation
**Mission Objective:** Apply unitary folding (U U† U) to scale circuit depth and measure noise sensitivity.

**Pedagogical Guidance:** Zero-Noise Extrapolation (ZNE) is the leading quantum error mitigation (QEM) technique used by IBM Quantum and scientific researchers today. Rather than correcting errors with millions of physical qubits, ZNE artificially amplifies noise by scale factors $\lambda \in [1, 3, 5]$ (via unitary folding $U \to U U^\dagger U$) and fits a curve to extrapolate back to zero noise $\lambda = 0$!

**Target Outcome:** `Observe folded circuit behavior under noise simulation`


```python
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator

qc = QuantumCircuit(1, 1)
qc.h(0); qc.h(0); qc.h(0)  # Folded gate: 3x noise depth
qc.measure(0, 0)

sim = AerSimulator()
counts = sim.run(transpile(qc, sim), shots=1024).result().get_counts()
print('Folded circuit measurement counts:', counts)
```

**Implementation Directives:**
